# Mini Project 1 To-Do list CLI application

```python
 # Createacommand-line application that allows users to add tasks, mark tasks as
 # completed, view pending tasks, and delete tasks.
 # ● Utilise basic data structures like lists or dictionaries to store tasks
```

```python
def menu():
    print("MENU\n")
    print("1.Add Task")
    print("2.Mark Task as Complete and delete")
    print("3.View Pending Tasks")
    print("4.Exit CLI")
```

```python
def add(tasks):
    task= input("Enter a task description")
    tasks = tasks.append(task)
    print(f'Task "{task}" added successfully\n')
```

```python
def mark(tasks):
    if not tasks:
        print("Task list is empty. Try adding a task.\n")
        return

    view(tasks)
    index = int(input("Enter the task index to delete: "))-1

    if 0<=index<len(tasks):
        removed= tasks.pop(index)
        print(f'Task "{removed}" has been removed\n')
    else:
        print("Invalid task index!!\n")
```

```python
def view(tasks):
    if not tasks:
        print("Task is not present\n")
        return
    for i,tasks in enumerate(tasks,start=1):
        print(f'{i}.{tasks}\n')
    
```

```python
def main():
    tasks =[]
    while True:
        menu()
        choice = int(input("Enter a choice: "))
        


        if(choice ==1):
            add(tasks)
        elif(choice ==2):
            mark(tasks)
        elif(choice ==3):
            view(tasks)
        elif(choice==4):
            print("Exiting...\n")
            break;
        else:
            print("Not valid input\n")

if __name__=="__main__":
    main()    
```

```python
#Second Lowest Grade (Nested List)
new_lst= []
count =5
while count>0:
    name,grade = input("Enter Student Name and Grade seperated by comma").split(',');
    new_lst.append([name,grade])
    count-=1
# print(new_lst)

values = sorted(set([item[1] for item in new_lst]))
second_lowest = values[1]
second_lowest_names = [item[0] for item in new_lst if item[1] == second_lowest]

second_lowest_names.sort()

print(second_lowest_names)
```

# Task 2: Number Guessing Game

```python
import random
rand_int= random.randint(1,101)
# print(rand_int)
count = 1
while True:
    val = int(input("Guess the number (Between 1 and 100): "))
    if val == rand_int:
        print(f"Correct ! You guessed the number in {count} attempt(s) \n")
        break
    elif val< rand_int:
        print("\nToo low!!. Try again!!\n")
        count+=1
    else:
        print("\nToo high!!. Try again!!\n")
        count+=1
```
